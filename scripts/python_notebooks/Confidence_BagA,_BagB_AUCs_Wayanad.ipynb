```python
import numpy as np, pandas as pd, glob
from sklearn.metrics import roc_auc_score

def boot(f, n=1000, seed=42):
    d = pd.read_csv(f).dropna(subset=['classification','is_target'])
    y, p = d['is_target'].values, d['classification'].values
    rng = np.random.default_rng(seed); s = []
    for _ in range(n):
        i = rng.choice(len(y), len(y), replace=True)
        if len(np.unique(y[i])) < 2:
            continue
        s.append(roc_auc_score(y[i], p[i]))
    s = np.array(s)
    return roc_auc_score(y, p), np.percentile(s, [2.5, 97.5]), (y==1).sum(), (y==0).sum()

for f in sorted(glob.glob('scored_*.csv')):
    auc, ci, npos, nneg = boot(f)
    print(f'{f:40s} pos={npos:3d} neg={nneg:3d} AUC={auc:.3f} [{ci[0]:.3f}, {ci[1]:.3f}]')
```
